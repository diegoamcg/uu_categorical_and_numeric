```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("./data/metabric.csv")

sns.ecdfplot(data=df, x="Overall Survival (Months)")
plt.xlabel("Overall Survival (Months)")
plt.ylabel("ECDF")
plt.title("ECDF of Overall Survival")
plt.show()
```

```python
sns.ecdfplot(data=df, x="Overall Survival (Months)", hue="Chemotherapy")
plt.xlabel("Overall Survival (Months)")
plt.ylabel("ECDF")
plt.title("ECDF of Overall Survival by Chemotherapy")
plt.show()
```
